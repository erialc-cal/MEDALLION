```python
import scanpy as sc
adata_ref = sc.datasets.pbmc3k_processed()
adata = sc.datasets.pbmc68k_reduced()
print(adata_ref.shape, adata.shape)
```

```python
var_names = adata_ref.var_names.intersection(adata.var_names) # get the shared features
adata_ref = adata_ref[:, var_names].copy()
adata = adata[:, var_names].copy()
```

```python
len(var_names) # 208 shared features
```

```python
import pandas as pd
df = pd.DataFrame(adata_ref.X)
df.columns = adata_ref.var_names
df.index = adata_ref.obs_names
df.to_csv('./data/pbmc3k_processed.csv')
```

```python
adata_ref.obs.to_csv('./data/pbmc3k_labels.csv')
```

```python
df = pd.DataFrame(adata.X)
df.columns = adata.var_names
df.index = adata.obs_names
df.to_csv('./data/pbmc68k_processed.csv')
```

```python
adata.obs.to_csv('./data/pbmc68k_labels.csv')
```
